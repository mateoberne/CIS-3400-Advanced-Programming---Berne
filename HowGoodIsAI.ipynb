# Step 1
Using an AI tool ChatGPT, we will enter:

Write a Python function named calculate_final_grade() that accepts a list of numeric quiz grades, drops the lowest quiz grade, and returns the average of the remaining grades.

ChatGPT_1.png

```python
def calculate_final_grade(grades):
    grades = grades.copy()
    grades.remove(min(grades))
    return sum(grades) / len(grades)
```

# Test the function with a simple list of numbers

```python
grades = [85, 92, 78, 90, 88]
print(calculate_final_grade(grades))
```

# Step 4: Run The Same Tests Again

Firstly, update the function

```python
def calculate_final_grade(grades):
    # Check for an empty list
    if len(grades) == 0:
        return "Error: The list cannot be empty."

    # Check for only one grade
    if len(grades) == 1:
        return "Error: You need at least two quiz grades to drop the lowest grade."

    # Check that all grades are numeric
    if not all(isinstance(grade, (int, float)) for grade in grades):
        return "Error: All quiz grades must be numbers."

    # Drop the lowest grade
    grades = grades.copy()
    grades.remove(min(grades))

    # Calculate the average
    return sum(grades) / len(grades)
```

# Now Run the Tests

```python
# Run to find out what happens if there's only one value in the list
print(calculate_final_grade([85]))
```
